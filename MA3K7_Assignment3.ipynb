```python
import numpy as np

heaps = list(map(int, input("Enter heaps as a tuple").strip("()").split(",")))  #player writes heaps as a tuple
heap_number = len(heaps)
Person = 0
Bot = 1

def nim_sum(heaps):   #function to calculate Nim sum
    total = heaps[0]   
    for i in range(1, len(heaps)):
        total = total ^ heaps[i]  #utilises Xor operation
    return total

def intelligent(heaps):   #strategy for the bot
    current_nim_sum = nim_sum(heaps)
    for i in range (heap_number):
        new_heap = heaps[i] ^ current_nim_sum   #Calculates what change needs to occur in a heap to make Nim sum 0 
        if new_heap < heaps[i]:   #Checks its subtracting stones
            take = heaps[i] - new_heap
            heaps[i] = new_heap
            return i, take   #returns the bots optimal move

    i = np.random.choice([i for i in range(heap_number) if heaps[i] > 0])   #this occurs if no optimal strategy is found, the Nim sum is already 0
    take = np.random.randint(1, heaps[i] + 1)   #chooses random valid values
    heaps[i] = heaps[i] - take
    return i, take

def game(player):   #players move
    if player == Person:
        print("Heaps:", heaps)
        heap = int(input("Which heap?")) - 1
        take = int(input("How many stones?"))
        heaps[heap] = heaps[heap] - take

        print("Heaps:", heaps)
        print("Nim sum:", nim_sum(heaps))

        if sum(heaps) == 0:   #winning criteria
            print("You won")
            return

        game(Bot)

    if player == Bot:
        heap, take = intelligent(heaps)   #uses bots strategy
        print("Bot took", take, "stones from heap", heap + 1)
        print("Heaps:", heaps)
        print("Nim sum:", nim_sum(heaps))

        if sum(heaps) == 0:   #bots winning criteria
            print("Bot won")
            return

        game(Person)

game(Person)
```
